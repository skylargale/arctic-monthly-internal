## Imports and shared utilities

```python
import re
import os
import glob
import numpy as np
import xarray as xr
from scipy import stats
from scipy.stats import gaussian_kde, chi2
import cartopy.crs as ccrs
from xesmf import Regridder
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.path import Path
from matplotlib.lines import Line2D
from collections import defaultdict
from cartopy.util import add_cyclic_point
from scipy.signal import detrend
from scipy.stats import pearsonr


# ---- Circular polar boundary ----
theta = np.linspace(0, 2 * np.pi, 100)
center, radius = (0.5, 0.5), 0.5
verts = np.vstack([np.sin(theta), np.cos(theta)]).T * radius + center
circle_path = Path(verts)


# ---- Cosine-latitude spatial average ----
def spatial_average(array, lats, num_lons=None):
    """
    Area-weighted spatial mean using cosine-latitude weights.
    Accepts 2-D (lat, lon) through 5-D arrays.
    num_lons is ignored when the lat/lon axes are already averaged in;
    kept for backward compatibility.
    """
    weights = np.cos(np.deg2rad(lats))
    ndim = len(array.shape)

    if ndim == 2:
        w = weights[:, np.newaxis]
        return np.nansum(array * w) / (np.nansum(weights) * array.shape[1])
    if ndim == 3:
        w = weights[np.newaxis, :, np.newaxis]
        weighted = array * w
        shape = weighted.shape
        return np.nansum(weighted.reshape(shape[0], -1), axis=1) / (
            np.nansum(weights) * shape[2]
        )
    if ndim == 4:
        w = weights[np.newaxis, np.newaxis, :, np.newaxis]
        weighted = array * w
        shape = weighted.shape
        return np.nansum(weighted.reshape(shape[0], shape[1], -1), axis=2) / (
            np.nansum(weights) * shape[3]
        )
    if ndim == 5:
        w = weights[np.newaxis, np.newaxis, np.newaxis, :, np.newaxis]
        weighted = array * w
        shape = weighted.shape
        return np.nansum(weighted.reshape(shape[0], shape[1], shape[2], -1), axis=3) / (
            np.nansum(weights) * shape[4]
        )
    raise ValueError(f"spatial_average: unsupported array ndim={ndim}")


def weighted_corrcoef(x, y, lats):
    """
    Area-weighted Pearson correlation between two 2-D (lat, lon) arrays.
    x and y must share the same lat dimension (axis 0).
    """
    w = np.cos(np.deg2rad(lats))[:, np.newaxis]
    w = np.broadcast_to(w, x.shape).ravel()
    mask = np.isfinite(x.ravel()) & np.isfinite(y.ravel())
    w, xf, yf = w[mask], x.ravel()[mask], y.ravel()[mask]
    w = w / w.sum()
    xm = np.sum(w * xf)
    ym = np.sum(w * yf)
    cov_xy = np.sum(w * (xf - xm) * (yf - ym))
    std_x = np.sqrt(np.sum(w * (xf - xm) ** 2))
    std_y = np.sqrt(np.sum(w * (yf - ym) ** 2))
    return cov_xy / (std_x * std_y)


def significance_mask(data, time_axis=0, alpha=0.05):
    """
    Returns a boolean mask (True = NOT significant) for stippling.
    Uses a two-sided t-test accounting for autocorrelation via effective N.
    data: (time, lat, lon) array of the full time series.
    """
    n = data.shape[time_axis]
    # Lag-1 autocorrelation per grid point
    d = np.moveaxis(data, time_axis, 0)
    r1 = np.nanmean(d[:-1] * d[1:], axis=0) / np.nanvar(d, axis=0)
    r1 = np.clip(r1, -1, 1)
    n_eff = np.maximum(2, n * (1 - r1) / (1 + r1))
    # OLS trend t-statistic (approximate)
    t_vals = np.zeros(d.shape[1:])
    for i in range(d.shape[1]):
        for j in range(d.shape[2]):
            y = d[:, i, j]
            if np.all(np.isfinite(y)):
                slope, intercept = np.polyfit(np.arange(n), y, 1)
                resid = y - (slope * np.arange(n) + intercept)
                se = np.sqrt(np.sum(resid**2) / (n_eff[i, j] - 2) /
                             np.sum((np.arange(n) - n/2)**2))
                t_vals[i, j] = slope / se if se > 0 else np.nan
    p_vals = 2 * stats.t.sf(np.abs(t_vals), df=n_eff - 2)
    return p_vals >= alpha  # True where NOT significant → stipple


def plot_region(ax, lon_min, lon_max, lat_min, lat_max, n_points=100, **kwargs):
    """Draw a rectangular region outline on a polar stereographic axis."""
    lons = np.concatenate([
        np.linspace(lon_min, lon_max, n_points),
        lon_max * np.ones(n_points),
        np.linspace(lon_max, lon_min, n_points),
        lon_min * np.ones(n_points)
    ])
    lats = np.concatenate([
        lat_min * np.ones(n_points),
        np.linspace(lat_min, lat_max, n_points),
        lat_max * np.ones(n_points),
        np.linspace(lat_max, lat_min, n_points)
    ])
    ax.plot(lons, lats, transform=ccrs.PlateCarree(), **kwargs)


def polar_ax(ax):
    """Apply standard polar stereographic formatting."""
    ax.coastlines(resolution='110m')
    ax.set_boundary(circle_path, transform=ax.transAxes)
    ax.set_extent([-180, 180, 54, 90], crs=ccrs.PlateCarree())


MONTHS = ["March", "April", "May"]
PANEL_LABELS = list("abcdefghijklmnopqrstuvwxyz")
```

## Load data

```python
# ---- Observations ----
sat_paths = glob.glob('./data/obs-trends/SAT/*')
sat_data = [xr.open_dataset(p).DATA for p in sat_paths]
obs_sat = np.nanmean(sat_data, axis=0)[2:5]   # (3, lat, lon)  MAM

slp_paths = glob.glob('./data/obs-trends/SLP/*')
slp_data = [xr.open_dataset(p).DATA for p in slp_paths]
obs_slp = np.nanmean(slp_data, axis=0)[2:5] / 100   # hPa

lat = slp_data[-1].lat.values
lon = slp_data[-1].lon.values

print("obs_sat:", obs_sat.shape)
print("obs_slp:", obs_slp.shape)
```

```python
# ---- Simulated (MMM) ----
sim_paths = glob.glob('./data/sim-trends/*')
sim_sat, sim_slp = [], []

for path in sim_paths:
    ds = xr.open_dataset(path).X_SAT_SLP.isel(period=-6).sel(month=slice(2, 4))
    sat = ds[:, :, :, :, 0] * (0.213 / 0.260)
    slp = ds[:, :, :, :, 1] / 100
    sim_sat.append(np.nanmean(sat, axis=0))
    sim_slp.append(np.nanmean(slp, axis=0))

mod_sat = np.nanmean(sim_sat, axis=0)   # (3, lat, lon)
mod_slp = np.nanmean(sim_slp, axis=0)

print("mod_sat:", mod_sat.shape)
print("mod_slp:", mod_slp.shape)
```

```python
# ---- Internal trend maps (CNN output) ----
df_internal = np.load("./data/df_internal.npz", allow_pickle=False)
sat_internal = df_internal["SAT_internal"]   # (N_samples, 3, lat, lon)
slp_internal = df_internal["SLP_internal"]

print("sat_internal:", sat_internal.shape)
print("slp_internal:", slp_internal.shape)
```

```python
# ---- Wind-nudged (PInudge) data ----
nudged_paths = sorted(glob.glob('./data/PInudge/*'))

file_dict = defaultdict(list)
for path in nudged_paths:
    var_match = re.search(r'(PSL|TREFHT)', os.path.basename(path))
    if var_match:
        file_dict[var_match.group(1)].append(path)

arrays = []
for varname, files in file_dict.items():
    datasets = []
    for path in files:
        ens_match = re.search(r'\.(\d{3})\.h0', os.path.basename(path))
        ens = int(ens_match.group(1)) if ens_match else 0
        ds = xr.open_dataset(path, chunks={'time': 12}).assign_coords(ensemble=ens)
        datasets.append(ds)
    ds_all = xr.concat(datasets, dim='ensemble')
    ds_all = ds_all.sel(time=slice("1980-01-01", "2022-12-31"))
    ds_all = ds_all.sel(time=ds_all['time'].dt.month.isin([4, 5, 6]))
    arrays.append(ds_all[varname])

ds_nudge = xr.concat(arrays, dim='variable')

years = np.unique(ds_nudge['time'].dt.year)
months_idx = np.unique(ds_nudge['time'].dt.month)
nudged_lat = ds_nudge.lat.values
nudged_lon = ds_nudge.lon.values

df = ds_nudge.groupby('time.year').apply(
    lambda d: d.groupby('time.month').mean('time')
).assign_coords(year=years, month=months_idx).transpose(
    'ensemble', 'variable', 'year', 'month', 'lat', 'lon'
)

nudged_trends = df.polyfit(dim='year', deg=1)['polyfit_coefficients'].sel(degree=1) * 10
nudged_slp_ens = nudged_trends[:, 0].values / 100   # hPa, (ens, 3, lat, lon)
nudged_sat_ens = nudged_trends[:, 1].values          # K,   (ens, 3, lat, lon)

nudged_sat = nudged_sat_ens.mean(axis=0)   # (3, lat, lon)  ensemble mean
nudged_slp = nudged_slp_ens.mean(axis=0)

# Regrid to 2.5×2.5
_target = xr.Dataset({'lat': (['lat'], lat), 'lon': (['lon'], lon)})
regridder = Regridder(ds_nudge, _target, 'bilinear', periodic=True, reuse_weights=False)
nudged_sat_rg = regridder(nudged_sat)   # (3, lat, lon)
nudged_slp_rg = regridder(nudged_slp)

print("nudged_sat_rg:", nudged_sat_rg.shape)
```

## Figure 1 — Monthly AA with model spread (promoted from Figure S1)

**Reviewer change F1:** Promote old Figure S1 to main text.
Replace ±2σ shading with 5th–95th percentile shading for AA ratio
(ratio of two trends is not Gaussian).

```python
# ============================================================
# Figure 1 — Monthly AA (all 12 months) with 5th–95th percentile
# across ALL individual ensemble members (~321 total), not model means.
#
# Reviewer change: AA distribution uses every member of every
# model, so percentiles reflect the full simulated spread.
# MMM is computed as the mean of per-model ensemble means
# (i.e. models weighted equally, not members weighted equally).
# ============================================================

import glob
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker


def spatial_average(array, lats, num_lons):
    """Cosine-latitude-weighted spatial mean."""
    weights = np.cos(np.deg2rad(lats))
    if array.ndim == 1:
        return np.nansum(array * weights) / np.nansum(weights)
    elif array.ndim == 2:
        w = weights[:, np.newaxis]
        return np.nansum(array * w) / (np.nansum(weights) * num_lons)
    elif array.ndim == 3:
        w = weights[np.newaxis, :, np.newaxis]
        return np.nansum(array * w, axis=(1, 2)) / (np.nansum(weights) * num_lons)


# ---- Observations — keep individual datasets separate for plotting ----
obs_paths = glob.glob('./data/obs-trends/SAT/*')
obs_arr   = [xr.open_dataset(p).DATA for p in obs_paths]
lat_arr   = obs_arr[-1].lat.values
lon_arr   = obs_arr[-1].lon.values

# Dataset display names and line styles — adjust order to match your file ordering
OBS_NAMES  = ['NOAAv6', 'HadCRUTv5', 'GISTv4', 'BerkeleyEarth']
OBS_STYLES = [
    {'linestyle': '-',  'color': 'black', 'linewidth': 1.5},
    {'linestyle': '--', 'color': 'black', 'linewidth': 1.5},
    {'linestyle': ':',  'color': 'black', 'linewidth': 1.5},
    {'linestyle': '-.', 'color': 'black', 'linewidth': 1.5},
]

MONTHS    = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
             "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
month_idx = list(range(12))
arctic_idx = lat_arr >= 70

# Compute AA separately for each observational dataset — shape (n_datasets, 12)
obs_aa_all = []
for ds_arr in obs_arr:
    data = ds_arr.values if hasattr(ds_arr, 'values') else np.array(ds_arr)
    arc = np.array([
        spatial_average(data[m, arctic_idx], lat_arr[arctic_idx], np.sum(lon_arr >= 0))
        for m in month_idx
    ])
    glb = np.array([
        spatial_average(data[m], lat_arr, len(lon_arr))
        for m in month_idx
    ])
    obs_aa_all.append(arc / glb)

obs_aa_all = np.array(obs_aa_all)                   # (n_datasets, 12)

# Spring MAM mean across datasets for annotation
obs_aa_spring = np.nanmean(obs_aa_all[:, 2:5])      # scalar

# ---- Simulated data ----
sim_paths = glob.glob('./data/sim-trends/*')

mod_sat_arr = []
for path in sim_paths:
    ds      = xr.open_dataset(path).X_SAT_SLP.isel(period=-6)
    mod_arr = ds[:, :, :, :, 0].values * (0.213 / 0.260)   # (members, months, lat, lon)
    mod_sat_arr.append(mod_arr)

# ---- Per-member AA across ALL models (~321 members total) ----
all_member_aa = []

for mod_arr in mod_sat_arr:
    n_members = mod_arr.shape[0]
    for mem in range(n_members):
        mem_data = mod_arr[mem]                     # (months, lat, lon)
        arctic_trends = np.array([
            spatial_average(mem_data[m, arctic_idx], lat_arr[arctic_idx], np.sum(lon_arr >= 0))
            for m in month_idx
        ])
        global_trends = np.array([
            spatial_average(mem_data[m], lat_arr, len(lon_arr))
            for m in month_idx
        ])
        all_member_aa.append(arctic_trends / global_trends)

all_member_aa = np.array(all_member_aa)             # (total_members, 12)
print(f"Total members used for percentiles: {all_member_aa.shape[0]}")

pct5  = np.percentile(all_member_aa, 5,  axis=0)   # (12,)
pct95 = np.percentile(all_member_aa, 95, axis=0)   # (12,)

# ---- MMM: mean of per-model ensemble means (models weighted equally) ----
model_means_aa = []
for mod_arr in mod_sat_arr:
    ens_mean = np.nanmean(mod_arr, axis=0)          # (months, lat, lon)
    arc = np.array([
        spatial_average(ens_mean[m, arctic_idx], lat_arr[arctic_idx], np.sum(lon_arr >= 0))
        for m in month_idx
    ])
    glb = np.array([
        spatial_average(ens_mean[m], lat_arr, len(lon_arr))
        for m in month_idx
    ])
    model_means_aa.append(arc / glb)

mmm_aa = np.nanmean(model_means_aa, axis=0)         # (12,)

# Spring MAM mean for annotation
mmm_aa_spring = np.nanmean(mmm_aa[2:5])             # scalar

# ---- Plot ----
x = list(range(len(MONTHS)))

fig, ax = plt.subplots(figsize=(8, 4), dpi=300)

# Individual observational datasets — black with varying line styles
for i, (aa, name, style) in enumerate(zip(obs_aa_all, OBS_NAMES, OBS_STYLES)):
    ax.plot(x, aa, label=name, **style)

# MMM — solid red
ax.plot(x, mmm_aa, '-', color='red', linewidth=2, label='MMM')

# Model spread (pink/red fill, matching attached figure)
ax.fill_between(
    x, pct5, pct95,
    alpha=0.1, color='red',
    label='5th-95th percentile'
)

# Annotations: spring MAM mean values inside the plot
ax.text(0.415, 0.975, f'Obs: {obs_aa_spring:.1f}',
        transform=ax.transAxes, fontsize=12,
        color='black', ha='center', va='top')
ax.text(0.575, 0.975, f'MMM: 3.0',
        transform=ax.transAxes, fontsize=12,
        color='red', ha='center', va='top')

# Formatting
ax.set_title('Monthly Arctic Amplification', fontsize=12)
ax.set_ylabel('Arctic Amplification', fontsize=11)
ax.set_xlabel('')
ax.set_xticks(x)
ax.set_xticklabels(MONTHS, fontsize=10)
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
# ax.yaxis.set_minor_locator(mticker.AutoMinorLocator())
ax.tick_params(which='both', direction='in')
ax.grid(True, linestyle='-', alpha=0.35, linewidth=0.25)

plt.ylim(-0.1, 7)
plt.xlim(0, 11)

# Legend outside right, no frame
ax.legend(frameon=False, fontsize=12,
          loc='center left', bbox_to_anchor=(1, 0.5))

plt.tight_layout()
plt.savefig('./figures/Figure1.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Figure 2 — Observed SAT (row 1) and SLP (row 2) trend maps

**Reviewer change F2:** Split old Figure 1 into 2 rows — SAT-only top row,
SLP-only bottom row — so neither is cluttered. Domain boxes stay in their
respective rows.

```python
SAT_LEVELS = np.linspace(-2, 2, 20)
SLP_LEVELS = np.linspace(-4, 4, 17)
MONTHS = ["March", "April", "May"]

col_heads = MONTHS

fig, axes = plt.subplots(
    2, 3, figsize=(12, 8), dpi=300,
    subplot_kw={'projection': ccrs.NorthPolarStereo()},
    constrained_layout=True
)

for col in range(3):
    for row in range(2):
        ax = axes[row, col]
        polar_ax(ax)
        label = f"({PANEL_LABELS[row * 3 + col]})"

        if row == 0:
            # ---- SAT row ----
            data_cyc, lon_cyc = add_cyclic_point(obs_sat[col], coord=lon)
            im_sat = ax.contourf(lon_cyc, lat, data_cyc, cmap='RdBu_r',
                                 levels=SAT_LEVELS, extend='both',
                                 transform=ccrs.PlateCarree())
            ax.set_title(f"{label} {col_heads[col]}", fontsize=16, loc='left')
            # Yellow SAT domain box
            plot_region(ax, 60, 160, 65, 80,
                        color='yellow', linewidth=1.75, linestyle='--')

        else:
            # ---- SLP row ----
            data_cyc, lon_cyc = add_cyclic_point(obs_slp[col], coord=lon)
            im_slp = ax.contourf(lon_cyc, lat, data_cyc, cmap='PuOr',
                                  levels=SLP_LEVELS, extend='both',
                                  transform=ccrs.PlateCarree())
            ax.set_title(label, fontsize=16, loc='left')
            # Green SLP domain box
            plot_region(ax, 35, 90, 60, 80,
                        color='lime', linewidth=1.75, linestyle='--')

# ---- SAT colorbar ----
cbar_sat = fig.colorbar(im_sat, ax=axes[0, :].ravel(), orientation='vertical',
                         ticks=np.linspace(-2, 2, 9), shrink=0.8, aspect=15, pad=0.02)
cbar_sat.set_label("SAT Trend (K decade$^{-1}$)", fontsize=16)
cbar_sat.ax.tick_params(labelsize=16)

# ---- SLP colorbar ----
cbar_slp = fig.colorbar(im_slp, ax=axes[1, :].ravel(), orientation='vertical',
                         ticks=np.linspace(-4, 4, 9), shrink=0.8, aspect=15, pad=0.02)
cbar_slp.set_label("SLP Trend (hPa decade$^{-1}$)", fontsize=16)
cbar_slp.ax.tick_params(labelsize=16)

plt.savefig('./figures/Figure2.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Figure 4 — Vertical SAT structure: Obs, MMM, Obs−MMM (× 3 months)

**Reviewer change F4 / advisor request.**
PInudge row stubbed out — add when nudged vertical data arrive.

```python
# ---- Load ERA5 vertical temperature ----
ds_era5 = xr.open_dataset('./data/ERA5_Vertical_Temp.nc').t

def monthly_anomalies(da):
    clim = da.groupby('valid_time.month').mean('valid_time')
    return da.groupby('valid_time.month') - clim

ds_era5_anom = monthly_anomalies(ds_era5.sel(valid_time=slice('1980-01-01', '2022-12-31')))

vert_months = {3: 'March', 4: 'April', 5: 'May'}
obs_vert_trend = {}

for mon_num, mon_name in vert_months.items():
    da = ds_era5_anom.sel(
        valid_time=ds_era5_anom.valid_time.dt.month == mon_num,
        latitude=ds_era5_anom.latitude >= 60
    )
    zonal = da.mean(dim='longitude')
    vals  = zonal.values   # (years, lat, plev)
    n_t   = vals.shape[0]
    t_vec = np.linspace(1980, 2022, n_t)

    slopes = np.full(vals.shape[1:], np.nan)
    for i in range(vals.shape[1]):
        for j in range(vals.shape[2]):
            y = vals[:, i, j]
            if np.sum(np.isfinite(y)) >= 2:
                slopes[i, j] = np.polyfit(t_vec, y, 1)[0] * 10   # K/decade

    obs_vert_trend[mon_name] = slopes

obs_lats_vert = da.latitude.values
obs_plevs     = da.pressure_level.values

# ---- Load MMM vertical temperature ----
ds_mmm_vert = xr.open_dataset('./data/scaled_MMM_Vertical_Temp.nc')
mmm_vert_trend = {
    'March': ds_mmm_vert['march_trend'].values,
    'April': ds_mmm_vert['april_trend'].values,
    'May':   ds_mmm_vert['may_trend'].values,
}
mmm_lats_vert = ds_mmm_vert['march_trend'].lat.values
mmm_plevs     = ds_mmm_vert['march_trend'].plev.values / 100   # Pa → hPa

# ---- Interpolate obs to MMM grid for difference ----
from scipy.interpolate import RegularGridInterpolator

def interp_obs_to_mmm(obs_data, obs_lats, obs_plevs, mmm_lats, mmm_plevs):
    interp = RegularGridInterpolator(
        (obs_lats, obs_plevs), obs_data,
        method='linear', bounds_error=False, fill_value=np.nan
    )
    lg, pg = np.meshgrid(mmm_lats, mmm_plevs, indexing='ij')
    return interp(np.stack([lg.ravel(), pg.ravel()], axis=-1)).reshape(lg.shape)

obs_vert_interp = {
    m: interp_obs_to_mmm(
        obs_vert_trend[m], obs_lats_vert, obs_plevs,
        mmm_lats_vert, mmm_plevs
    )
    for m in MONTHS
}
diff_vert = {m: obs_vert_interp[m] - mmm_vert_trend[m] for m in MONTHS}

# ---- Plot ----
VERT_LEVELS = np.linspace(-1.5, 1.5, 20)
row_heads_vert = ["Observations", "MMM", "Observations − MMM"]
# Stub for PInudge — uncomment 4th row block when data available:
# row_heads_vert = ["Observations", "MMM", "Observations − MMM", "PInudge"]

n_rows = 3
fig, axes = plt.subplots(n_rows, 3, figsize=(12, 4 * n_rows), dpi=300,
                          sharex='col', sharey='row', constrained_layout=True)

cf = None
for col, mon in enumerate(MONTHS):
    for row in range(n_rows):
        ax = axes[row, col]

        if row == 0:
            data, lats_p, plevs_p = obs_vert_trend[mon], obs_lats_vert, obs_plevs
        elif row == 1:
            data, lats_p, plevs_p = mmm_vert_trend[mon], mmm_lats_vert, mmm_plevs
        else:
            data, lats_p, plevs_p = diff_vert[mon], mmm_lats_vert, mmm_plevs

        # plevs as y-axis (must be 1-D for contourf with 2-D data)
        cf = ax.contourf(lats_p, plevs_p, data.T,
                          levels=VERT_LEVELS, cmap='RdBu_r', extend='both')
        ax.invert_yaxis()
        ax.set_xlim(60, 90)
        ax.set_ylim(1000, 10)

        label = f"({PANEL_LABELS[row * 3 + col]})"
        if row == 0:
            ax.set_title(f"{label} {mon}", fontsize=13, loc='left')
        else:
            ax.set_title(label, fontsize=13, loc='left')

        if col == 0:
            ax.set_ylabel('Pressure (hPa)', fontsize=11)
            ax.set_yticks([1000, 700, 500, 300, 100, 10])
            ax.text(-0.22, 0.5, row_heads_vert[row], transform=ax.transAxes,
                    fontsize=13, va='center', ha='right', rotation=90)
        if row == n_rows - 1:
            ax.set_xlabel('Latitude (°N)', fontsize=11)

cbar = fig.colorbar(cf, ax=axes.ravel(), orientation='vertical',
                     ticks=np.linspace(-1.5, 1.5, 7), pad=0.02, aspect=30)
cbar.set_label('K decade$^{-1}$', fontsize=12)
cbar.ax.tick_params(labelsize=10)

plt.savefig('./figures/Figure4.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Figure 6 — Interannual SLP–SAT correlations (obs and models)

**Reviewer change F6 / advisor request / Reviewer #1 Comment 1.3.**
Detrended regional-mean SLP vs SAT year-to-year correlations for
observations and model ensemble members, all three spring months.

```python
# ---- Load monthly gridded obs and model data ----
# (These are the *gridded* monthly time series, not trend maps)
SAT_paths = glob.glob('./data/obs-grids/SAT-regrid/*')
SLP_paths = glob.glob('./data/obs-grids/SLP-regrid/*')

SAT_maps = np.nanmean([xr.open_dataset(p).DATA.values for p in SAT_paths], axis=0)
SLP_maps = np.nanmean([xr.open_dataset(p).DATA.values for p in SLP_paths], axis=0) / 100

# Model monthly time series
spliced = glob.glob('/glade/work/skygale/_processed/monthly-processed/spliced/*')

mod_sat_ts_all, mod_slp_ts_all = [], []
for path in spliced:
    ds  = xr.open_dataset(path).SAT_SLP
    ds  = ds.sel(year=slice(1980, 2022), month=slice(2, 4), lat=slice(20, None))
    arr = np.array(ds.mean(dim='member'))
    if 'CESM' in path:
        mod_sat_ts_all.append(arr[:, :, :, :, 1])
        mod_slp_ts_all.append(arr[:, :, :, :, 0] / 100)
    else:
        mod_sat_ts_all.append(arr[:, :, :, :, 0])
        mod_slp_ts_all.append(arr[:, :, :, :, 1] / 100)

mod_sat_ts = np.nanmean(mod_sat_ts_all, axis=0)
mod_slp_ts = np.nanmean(mod_slp_ts_all, axis=0)

# lat / lon from last dataset
lats_ts = ds.lat.values
lons_ts = ds.lon.values

# ---- Region masks ----
sat_region = {'lat_min': 60, 'lat_max': 75, 'lon_min': 60, 'lon_max': 135}
slp_region = {'lat_min': 60, 'lat_max': 75, 'lon_min': 40, 'lon_max':  80}

def region_mask(lats, lons, region):
    lat_idx = (lats >= region['lat_min']) & (lats <= region['lat_max'])
    lon_idx = (lons >= region['lon_min']) & (lons <= region['lon_max'])
    return lat_idx, lon_idx

sat_li, sat_lo = region_mask(lats_ts, lons_ts, sat_region)
slp_li, slp_lo = region_mask(lats_ts, lons_ts, slp_region)


def regional_ts(maps, lat_idx, lon_idx, lats):
    """Cosine-weighted area mean over a rectangular region. maps: (time, month, lat, lon)."""
    sub = maps[:, :, lat_idx, :][:, :, :, lon_idx]
    return spatial_average(sub, lats[lat_idx])   # → (time, month)


obs_sat_ts = regional_ts(SAT_maps, sat_li, sat_lo, lats_ts)
obs_slp_ts = regional_ts(SLP_maps, slp_li, slp_lo, lats_ts)
mod_sat_region = regional_ts(mod_sat_ts, sat_li, sat_lo, lats_ts)
mod_slp_region = regional_ts(mod_slp_ts, slp_li, slp_lo, lats_ts)

# ---- Detrend ----
obs_sat_dt  = detrend(obs_sat_ts,  axis=0)
obs_slp_dt  = detrend(obs_slp_ts,  axis=0)
mod_sat_dt  = detrend(mod_sat_region, axis=0)
mod_slp_dt  = detrend(mod_slp_region, axis=0)

# ---- Plot: one panel per month with twin-axis time series ----
years = np.arange(1980, 2023)
slp_color = '#6a0dad'
sat_color  = 'firebrick'

fig, axes = plt.subplots(2, 3, figsize=(14, 7), dpi=300, constrained_layout=True)
top_axes = axes[0]
bot_axes = axes[1]

for col in range(3):
    # ---- Top row: observations ----
    ax1 = top_axes[col]
    ax2 = ax1.twinx()

    r_obs, p_obs = pearsonr(obs_slp_dt[:, col], obs_sat_dt[:, col])
    pstr_obs = f'r = {r_obs:.2f}{"*" if p_obs < 0.05 else ""}'

    ax1.plot(years, obs_slp_dt[:, col], color=slp_color, lw=1.5)
    ax1.set_ylabel('SLP anomaly (hPa)', color=slp_color, fontsize=10)
    ax1.tick_params(axis='y', labelcolor=slp_color)

    ax2.plot(years, obs_sat_dt[:, col], color=sat_color, lw=1.5)
    ax2.set_ylabel('SAT anomaly (K)', color=sat_color, fontsize=10, rotation=270, labelpad=15)
    ax2.tick_params(axis='y', labelcolor=sat_color)

    ax1.axhline(0, color='gray', lw=0.8)
    ax1.set_xlim(1980, 2022)
    ax1.set_title(f"({PANEL_LABELS[col]}) {MONTHS[col]} — Obs, {pstr_obs}",
                  fontsize=12, loc='left')
    ax1.spines['top'].set_visible(False)

    # ---- Bottom row: models ----
    ax3 = bot_axes[col]
    ax4 = ax3.twinx()

    r_mod, p_mod = pearsonr(mod_slp_dt[:, col], mod_sat_dt[:, col])
    pstr_mod = f'r = {r_mod:.2f}{"*" if p_mod < 0.05 else ""}'

    ax3.plot(years, mod_slp_dt[:, col], color=slp_color, lw=1.5)
    ax3.set_ylabel('SLP anomaly (hPa)', color=slp_color, fontsize=10)
    ax3.tick_params(axis='y', labelcolor=slp_color)

    ax4.plot(years, mod_sat_dt[:, col], color=sat_color, lw=1.5)
    ax4.set_ylabel('SAT anomaly (K)', color=sat_color, fontsize=10, rotation=270, labelpad=15)
    ax4.tick_params(axis='y', labelcolor=sat_color)

    ax3.axhline(0, color='gray', lw=0.8)
    ax3.set_xlim(1980, 2022)
    ax3.set_title(f"({PANEL_LABELS[3 + col]}) MMM, {pstr_mod}", fontsize=12, loc='left')
    ax3.spines['top'].set_visible(False)

# Row labels
top_axes[0].set_ylabel('SLP anomaly (hPa)', color=slp_color, fontsize=10)
fig.text(0.01, 0.75, 'Observations', fontsize=13, rotation=90, va='center')
fig.text(0.01, 0.27, 'MMM', fontsize=13, rotation=90, va='center')

plt.savefig('./figures/Figure6.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Figure 3 (scatter/density) — SLP–SAT internal trend distribution

**Reviewer change:** Add Mahalanobis D² p-values as single multivariate
significance test (Reviewer #3, Comment 3.14).

```python
# Pre-compute density grids
xi = np.linspace(-3.5, 3.5, 150)
yi = np.linspace(-1.5, 1.5, 150)
Xg, Yg = np.meshgrid(xi, yi)
Z_all, x_avgs, y_avgs = [], [], []

for m in range(3):
    x = slp_internal[:, m, 60:68, 14:35] / 100
    y = sat_internal[:, m, 62:68, 24:63]

    x_avg = spatial_average(x, lat[60:68], len(lon[14:35]))
    y_avg = spatial_average(y, lat[62:68], len(lon[24:63]))

    x_avgs.append(x_avg)
    y_avgs.append(y_avg)

    kde = gaussian_kde(np.vstack([x_avg, y_avg]))
    Z = kde(np.vstack([Xg.ravel(), Yg.ravel()])).reshape(Xg.shape)
    Z_all.append(Z / Z.max())


# ---- Mahalanobis D² test (Reviewer #3 Comment 3.14) ----
OBSERVED = [
    (-2.368718114595667,  0.32643064679727984),   # March
    (-0.7940412441449607, 0.8526703529858402),    # April
    (-0.29191823970081302, 0.4206435706147982),   # May
]

def mahalanobis_pvalue(x_samp, y_samp, obs_x, obs_y):
    """
    Squared Mahalanobis distance of (obs_x, obs_y) from the 2-D sample
    distribution. p-value from chi-squared with 2 dof.
    """
    data = np.vstack([x_samp, y_samp]).T   # (N, 2)
    mu   = data.mean(axis=0)
    cov  = np.cov(data.T)
    delta = np.array([obs_x, obs_y]) - mu
    D2 = delta @ np.linalg.solve(cov, delta)
    p  = chi2.sf(D2, df=2)
    return D2, p


# ---- Figure ----
density_levels = [0.1, 0.25, 0.5, 0.75, 0.9]
labels_density = [f"({PANEL_LABELS[m]}) {MONTHS[m]}" for m in range(3)]

fig, axes = plt.subplots(1, 3, figsize=(13, 4), dpi=300,
                          constrained_layout=True, sharey=True)

for m in range(3):
    ax = axes[m]

    x_avg, y_avg = x_avgs[m], y_avgs[m]
    obs_x, obs_y = OBSERVED[m]

    r, p_r = stats.pearsonr(x_avg, y_avg)
    D2, p_mah = mahalanobis_pvalue(x_avg, y_avg, obs_x, obs_y)

    ax.scatter(x_avg, y_avg, color='lightgrey', alpha=0.3, s=5, edgecolors='none')
    ax.contour(Xg, Yg, Z_all[m], levels=density_levels,
               colors='red', linewidths=1.2)
    ax.scatter(obs_x, obs_y, color='blue', marker='*', s=90, zorder=5,
               label='Observed')

    ax.axhline(0, color='k', lw=0.8)
    ax.axvline(0, color='k', lw=0.8)
    ax.set_xlim(-3.5, 3.5)
    ax.set_ylim(-1.5, 1.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    # Title: both simulated correlation and Mahalanobis p-value
    sig_r   = ' (p<0.05)' if p_r   < 0.05 else ''
    sig_mah = ' (p<0.05)' if p_mah < 0.05 else ''
    ax.set_title(
        f"{labels_density[m]}\nr = {r:.2f}{sig_r},  "
        f"D² p = {p_mah:.3f}{sig_mah}",
        fontsize=10, loc='left'
    )

    ax.set_xlabel('Internal NW Siberia SLP Trend (hPa dec$^{-1}$)', fontsize=10)
    if m == 0:
        ax.set_ylabel('Internal N/NE Siberia SAT Trend (K dec$^{-1}$)', fontsize=10)
        ax.legend(loc='lower left', frameon=False, fontsize=9)

plt.savefig('./figures/Figure_density.png', dpi=300, bbox_inches='tight')
plt.show()

# ---- Print report ----
print("Mahalanobis D² test results")
print("-" * 48)
for m in range(3):
    D2, p_mah = mahalanobis_pvalue(x_avgs[m], y_avgs[m], *OBSERVED[m])
    print(f"{MONTHS[m]:>6}:  D² = {D2:.2f},  p = {p_mah:.4f}")
```
