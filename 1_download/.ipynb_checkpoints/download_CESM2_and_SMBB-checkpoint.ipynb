```python
import numpy as np
import xarray as xr
import glob
import xesmf as xe
import netCDF4 as nc
import warnings
warnings.filterwarnings("ignore")
```

```python
def tas_append_data(list):
    appended_array = []
    for file in list:
        ds = xr.open_dataset(file)['TREFHT']
        appended_array.append(ds)
    return appended_array


def psl_append_data(list):
    appended_array = []
    for file in list:
        ds = xr.open_dataset(file)['PSL']
        appended_array.append(ds)
    return appended_array


def saver(shape, combined_data, type):
    lons = np.arange(1.25, 358.751, 2.5)
    lats = np.arange(-88.75, 88.751, 2.5)
    years = np.arange(1851, 1851 + shape[1])
    months = np.arange(0, 12)

    filename = '/glade/work/skygale/mydata/ssp_processed/'

    # Create a netcdf file
    if type == 'normal':
        fileName = filename + 'CESM2.nc'
    elif type == 'smbb':
        fileName = filename + 'CESM2_SMBB.nc'

    dataset = nc.Dataset(fileName, 'w', format='NETCDF4')

    # Define the dimensions
    dataset.createDimension('member', shape[0])  # 50
    dataset.createDimension('year', shape[1])    # 250
    dataset.createDimension('month', shape[2])   # 12
    dataset.createDimension('lat', shape[3])     # 72
    dataset.createDimension('lon', shape[4])     # 144
    dataset.createDimension('data', shape[5])    # 2

    # Define the variables
    var = dataset.createVariable('SAT_SLP', np.float32, ('member', 'year', 'month', 'lat', 'lon', 'data'))
    var[:] = combined_data
    dataset.close()

    # Open the NetCDF file in append mode
    dataset = nc.Dataset(fileName, 'a')

    members_var = dataset.createVariable('member', 'f8', ('member',))
    members_var[:] = np.arange(1, 1+shape[0])

    year_var = dataset.createVariable('year', 'f8', ('year',))
    year_var[:] = years

    months_var = dataset.createVariable('month', 'f8', ('month',))
    months_var[:] = months

    lat_var = dataset.createVariable('lat', 'f8', ('lat',))
    lat_var[:] = lats

    lon_var = dataset.createVariable('lon', 'f8', ('lon',))
    lon_var[:] = lons

    tas_and_psl_var = dataset.createVariable('data', 'f8', ('data',))
    tas_and_psl_var[:] = [1, 2]

    dataset.close()
```

```python
# define CESM2 specific realization and member labels
realizations = ['-1001.', '-1011.', '-1021.', '-1031.', '-1041.', '-1051.',
                '-1061.', '-1071.', '-1081.', '-1091.', '-1101.', '-1111.',
                '-1121.', '-1131.', '-1141.', '-1151.', '-1161.', '-1171.',
                '-1181.', '-1191.', '-1201.', '-1211.', '-1221.', '-1231.',
                '-1241.', '-1251.', '-1261.', '-1271.', '-1281.', '-1291.',
                '-1301.']

members = ['.001.', '.002.', '.003.', '.004.', '.005.', '.006.', '.007.',
           '.008.', '.009.', '.010.', '.011.', '.012.', '.013.', '.014.',
           '.015.', '.016.', '.017.', '.018.', '.019.', '.020.']
```

```python
# sort all data into appropriate lists
tas_paths = glob.glob('/glade/campaign/cgd/cesm/CESM2-LE/timeseries/atm/proc/tseries/month_1/TREFHT/*')
psl_paths = glob.glob('/glade/campaign/cgd/cesm/CESM2-LE/timeseries/atm/proc/tseries/month_1/PSL/*')

tas_hist = []
tas_hist_smbb = []
tas_ssp = []
tas_ssp_smbb = []

psl_hist = []
psl_hist_smbb = []
psl_ssp = []
psl_ssp_smbb = []

for path in tas_paths:
    for realization in realizations:
        for member in members:
            if realization in path:
                if member in path:
                    if 'BHISTcmip6' in path:
                        tas_hist.append(path)
                    if 'BHISTsmbb' in path:
                        tas_hist_smbb.append(path)
                    if 'BSSP370cmip6' in path:
                        tas_ssp.append(path)
                    if 'BSSP370smbb' in path:
                        tas_ssp_smbb.append(path)

for path in psl_paths:
    for realization in realizations:
        for member in members:
            if realization in path:
                if member in path:
                    if 'BHISTcmip6' in path:
                        psl_hist.append(path)
                    if 'BHISTsmbb' in path:
                        psl_hist_smbb.append(path)
                    if 'BSSP370cmip6' in path:
                        psl_ssp.append(path)
                    if 'BSSP370smbb' in path:
                        psl_ssp_smbb.append(path)

# sort by time and member
tas_hist = sorted(tas_hist)
tas_hist_smbb = sorted(tas_hist_smbb)
tas_ssp = sorted(tas_ssp)
tas_ssp_smbb = sorted(tas_ssp_smbb)

psl_hist = sorted(psl_hist)
psl_hist_smbb = sorted(psl_hist_smbb)
psl_ssp = sorted(psl_ssp)
psl_ssp_smbb = sorted(psl_ssp_smbb)

# print lengths
print('TAS:', len(tas_hist), len(tas_hist_smbb), len(tas_ssp), len(tas_ssp_smbb))
print('PSL:', len(psl_hist), len(psl_hist_smbb), len(psl_ssp), len(psl_ssp_smbb))
```

```python
tas_all_members = []
tas_smbb_all_members = []
psl_all_members = []
psl_smbb_all_members = []
```

```python
for i in range(50):
    # get each member hist and ssp
    tas_hist_mem = tas_hist[i*17:i*17+17]
    tas_ssp_mem = tas_ssp[i*9:i*9+9]
    tas_hist_smbb_mem = tas_hist_smbb[i*17:i*17+17]
    tas_ssp_smbb_mem = tas_ssp_smbb[i*9:i*9+9]
    psl_hist_mem = psl_hist[i*17:i*17+17]
    psl_ssp_mem = psl_ssp[i*9:i*9+9]
    psl_hist_smbb_mem = psl_hist_smbb[i*17:i*17+17]
    psl_ssp_smbb_mem = psl_ssp_smbb[i*9:i*9+9]

    # add them into one array for the member
    tas_add = tas_hist_mem + tas_ssp_mem
    tas_smbb_add = tas_hist_smbb_mem + tas_ssp_smbb_mem
    psl_add = psl_hist_mem + psl_ssp_mem
    psl_smbb_add = psl_hist_smbb_mem + psl_ssp_smbb_mem

    # append the xr data
    tas_appended = tas_append_data(tas_add)
    tas_smbb_appended = tas_append_data(tas_smbb_add)
    psl_appended = psl_append_data(psl_add)
    psl_smbb_appended = psl_append_data(psl_smbb_add)

    # concat along time dim
    tas_xr = xr.concat(tas_appended, dim='time')
    tas_smbb_xr = xr.concat(tas_smbb_appended, dim='time')
    psl_xr = xr.concat(psl_appended, dim='time')
    psl_smbb_xr = xr.concat(psl_smbb_appended, dim='time')

    # cut off unwanted time
    tas_cut = tas_xr[11:-1]
    tas_smbb_cut = tas_smbb_xr[11:-1]
    psl_cut = psl_xr[11:-1]
    psl_smbb_cut = psl_smbb_xr[11:-1]

    # reshape into years, months
    s = np.shape(psl_smbb_cut)
    tas_rs = np.reshape(tas_cut.to_numpy(), (int(s[0]/12), 12, s[1], s[2]))
    tas_smbb_rs = np.reshape(tas_smbb_cut.to_numpy(), (int(s[0]/12), 12, s[1], s[2]))
    psl_rs = np.reshape(psl_cut.to_numpy(), (int(s[0]/12), 12, s[1], s[2]))
    psl_smbb_rs = np.reshape(psl_smbb_cut.to_numpy(), (int(s[0]/12), 12, s[1], s[2]))

    # append to all member array
    tas_all_members.append(tas_rs)
    tas_smbb_all_members.append(tas_smbb_rs)
    psl_all_members.append(psl_rs)
    psl_smbb_all_members.append(psl_smbb_rs)

    print('Done with Member #', i+1)

print('Done!')
```

```python
# convert back into DataArray
member = np.arange(1, 51)
year = np.arange(1851, 2101)
month = np.arange(0, 12)
lats = psl_smbb_xr.lat.values
lons = psl_smbb_xr.lon.values
```

```python
tas_xr_array = xr.DataArray(tas_all_members,
                            coords={'member': member,
                                    'year': year,
                                    'month': month,
                                    'lat': lats,
                                    'lon': lons},
                            dims=['member', 'year', 'month', 'lat', 'lon'])
```

```python
tas_smbb_xr_array = xr.DataArray(tas_smbb_all_members,
                                 coords={'member': member,
                                         'year': year,
                                         'month': month,
                                         'lat': lats,
                                         'lon': lons},
                                 dims=['member', 'year', 'month', 'lat', 'lon'])
```

```python
psl_xr_array = xr.DataArray(psl_all_members,
                            coords={'member': member,
                                    'year': year,
                                    'month': month,
                                    'lat': lats,
                                    'lon': lons},
                            dims=['member', 'year', 'month', 'lat', 'lon'])
```

```python
psl_smbb_xr_array = xr.DataArray(psl_smbb_all_members,
                                 coords={'member': member,
                                         'year': year,
                                         'month': month,
                                         'lat': lats,
                                         'lon': lons},
                                 dims=['member', 'year', 'month', 'lat', 'lon'])
```

```python
# regrid
InputGrid = {"lon": lons, "lat": lats}
OutputGrid = {"lon": np.arange(1.25, 358.751, 2.5), "lat": np.arange(-88.75, 88.751, 2.5)}
regridder = xe.Regridder(InputGrid, OutputGrid, "bilinear", periodic=True)
```

```python
tas_regrid = regridder(tas_xr_array)
```

```python
tas_smbb_regrid = regridder(tas_smbb_xr_array)
```

```python
psl_regrid = regridder(psl_xr_array)
```

```python
psl_smbb_regrid = regridder(psl_smbb_xr_array)
```

```python
# combine TAS and PSL data into single array
combined = [psl_regrid, tas_regrid]
combined = np.moveaxis(combined, 0, -1)
shape = np.shape(combined)
```

```python
combined_smbb = [psl_smbb_regrid, tas_smbb_regrid]
combined_smbb = np.moveaxis(combined_smbb, 0, -1)
shape_smbb = np.shape(combined_smbb)
```

```python
shape, shape_smbb  # (50, 200, 12, 72, 144, 2)
```

```python
# save CESM2 as .nc file in mydata
saver(shape, combined, 'normal')
```

```python
# save CESM2_SMBB as .nc file in mydata
saver(shape_smbb, combined_smbb, 'smbb')
```
