```python
import re
import os
import glob
import numpy as np
import xesmf as xe
import xarray as xr
import pandas as pd
from scipy import stats
import cartopy.crs as ccrs
from xesmf import Regridder
from pyproj import Transformer
import matplotlib.pyplot as plt
from matplotlib.path import Path
import matplotlib.lines as mlines
import cartopy.feature as cfeature
from collections import defaultdict
import matplotlib.colors as mcolors
from matplotlib.lines import Line2D
from scipy.stats import gaussian_kde
import matplotlib.patches as mpatches
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
from cartopy.util import add_cyclic_point
from sklearn.linear_model import LinearRegression

# Functions
def spatial_average(array, lats, num_lons):
    weights = np.cos(np.deg2rad(lats))
    if len(array.shape) == 2:
        weights_array = weights[:, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0]*shape[1]))
        weighted_sum = np.nansum(weighted_reshaped, axis=0)
    if len(array.shape) == 3:
        weights_array = weights[np.newaxis, :, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0], shape[1]*shape[2]))
        weighted_sum = np.nansum(weighted_reshaped, axis=1)
    if len(array.shape) == 4:
        weights_array = weights[np.newaxis, np.newaxis, :, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0], shape[1], shape[2]*shape[3]))
        weighted_sum = np.nansum(weighted_reshaped, axis=2)
    if len(array.shape) == 5:
        weights_array = weights[np.newaxis, np.newaxis, np.newaxis, :, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0], shape[1], shape[2], shape[3]*shape[4]))
        weighted_sum = np.nansum(weighted_reshaped, axis=3)
    denom = np.nansum(weights) * num_lons
    return np.divide(weighted_sum, denom)
    
    
def plot_region(ax, lon_min, lon_max, lat_min, lat_max, n_points=100, **kwargs):
    lons = np.concatenate([
        np.linspace(lon_min, lon_max, n_points),
        lon_max*np.ones(n_points),
        np.linspace(lon_max, lon_min, n_points),
        lon_min*np.ones(n_points)
    ])
    lats = np.concatenate([
        lat_min*np.ones(n_points),
        np.linspace(lat_min, lat_max, n_points),
        lat_max*np.ones(n_points),
        np.linspace(lat_max, lat_min, n_points)
    ])
    ax.plot(lons, lats, transform=ccrs.PlateCarree(), **kwargs)


# Circular boundary for clipping
theta = np.linspace(0, 2 * np.pi, 100)
center = (0.5, 0.5)
radius = 0.5  # in axes coords
verts = np.vstack([np.sin(theta), np.cos(theta)]).T * radius + center
circle_path = Path(verts)
```

### Load data

```python
# ---- Observations ----

# Surface air temperature (K)
sat_paths = glob.glob('./data/obs-trends/SAT/*')

sat_data = []
for path in sat_paths:
    ds = xr.open_dataset(path).DATA
    sat_data.append(ds)

obs_sat = np.nanmean(sat_data, axis=0)[2:5]

# Sea level pressure (hPa)
slp_paths = glob.glob('./data/obs-trends/SLP/*')

slp_data = []
for path in slp_paths:
    ds = xr.open_dataset(path).DATA
    slp_data.append(ds)
    
obs_slp = np.nanmean(slp_data, axis=0)[2:5] / 100

# lat, lon
lat = ds.lat.values
lon = ds.lon.values

# Check shapes
print('obs_sat:', np.shape(obs_sat))
print('obs_slp:', np.shape(obs_slp))
```

```python
# ----- Simulated Data -----
sim_paths = glob.glob('./data/sim-trends/*')

sim_sat, sim_slp = [], []
for path in sim_paths:

    # Select MAM, 1980-2022, scaled for forcing bias
    ds  = xr.open_dataset(path).X_SAT_SLP.isel(period=-2).sel(month=slice(2, 4))[:, :, :, :, 0] * (0.213/0.260)
    ds_ = xr.open_dataset(path).X_SAT_SLP.isel(period=-2).sel(month=slice(2, 4))[:, :, :, :, 1]

    # Append ensemble mean
    sim_sat.append(np.nanmean(ds, axis=0))
    sim_slp.append(np.nanmean(ds_, axis=0))

# MMM
mod_sat = np.nanmean(sim_sat, axis=0)
mod_slp = np.nanmean(sim_slp, axis=0)

# Check shapes
print('mod_sat:', np.shape(mod_sat))
print('mod_slp:', np.shape(mod_slp))
```

```python
# ----- Internal Trend Maps -----
df_internal = np.load("./data/df_internal.npz", allow_pickle=False)

sat_internal = df_internal["SAT_internal"]   # shape: (N, 72, 144)
slp_internal = df_internal["SLP_internal"]

# Check shape
print('sat_internal:', np.shape(sat_internal))
print('slp_internal:', np.shape(slp_internal))
```

```python
# ---- Wind-Nudged Data ----
nudged_paths = sorted(glob.glob('./data/PInudge/*'))

# Group files by variable name
file_dict = defaultdict(list)
for path in nudged_paths:
    fname = os.path.basename(path)
    var_match = re.search(r'(PSL|TREFHT)', fname)
    if var_match:
        varname = var_match.group(1)
        file_dict[varname].append(path)

# Put all processed data together
arrays = []
for varname, files in file_dict.items():
    # print(f"Processing {varname} from {len(files)} file(s)...")

    # Each file is a different ensemble member
    datasets = []
    for path in files:
        ens_match = re.search(r'\.(\d{3})\.h0', os.path.basename(path))
        ens = ens_match.group(1) if ens_match else '000'
        ds = xr.open_dataset(path, chunks={'time': 12})
        ds = ds.assign_coords(ensemble=int(ens))
        datasets.append(ds)

    # Combine along a new ensemble dimension
    ds = xr.concat(datasets, dim='ensemble')

    # Select March–May, 1980–2022
    ds = ds.sel(time=slice("1980-01-01", "2022-12-31"))
    ds = ds.sel(time=ds['time'].dt.month.isin([4, 5, 6]))

    # Append to list
    arrays.append(ds[varname])

# Combine all variables
ds = xr.concat(arrays, dim='variable')

# Create year and month coordinates
years = np.unique(ds['time'].dt.year)
months = np.unique(ds['time'].dt.month)

# Group by year and month
df = ds.groupby('time.year').apply(
    lambda ds_year: ds_year.groupby('time.month').mean('time')
)

# Assign year and month coordinates
df = df.assign_coords(
    year=years,
    month=months
).transpose('ensemble', 'variable', 'year', 'month', 'lat', 'lon')
df.name = 'nudged_variables'

# Compute the linear trend for each variable, month, lat, and lon per decade
nudged_trends = df.polyfit(dim='year', deg=1)['polyfit_coefficients'].sel(degree=1) * 10
nudged_trends
```

```python
# Get separate numpy arrays
nudged_slp_ens = nudged_trends[:, 0].values / 100  # hPa
nudged_sat_ens = nudged_trends[:, 1].values

# Array with separate ensemble members (native grid)
# (ens, months, lat, lon)
print('nudged_sat_ens:', np.shape(nudged_sat_ens))

# Array with ensemble mean (native grid)
nudged_sat = np.mean(nudged_sat_ens, axis=0)
nudged_slp = np.mean(nudged_slp_ens, axis=0)
print('nudged_sat:    ', np.shape(nudged_sat))

# Array with ensemble mean (regridded to 2.5 x 2.5 deg)
regridder = Regridder(ds, xr.Dataset({'lat': (['lat'], lat), 'lon': (['lon'], lon)}), 'bilinear', reuse_weights=False) 
nudged_sat_rg = regridder(nudged_sat)
print('nudged_sat_rg: ', np.shape(nudged_sat_rg))
```

#### Correlation Check (Fig 7e)

```python
# Change nudged data to '12-31-2023' for this

# Load ERA5 data
ERA5 = xr.open_dataset('./data/ERA5_t2m_MAM_1980_2023_hires.nc').t2m

# Add numeric year coordinate
ERA5 = ERA5.assign_coords(
    year=ERA5["valid_time"].dt.year
)

ERA5 = ERA5.sel(valid_time=slice("1980-01-01", "2022-12-31"))

# Monthly climatology (mean for each calendar month)
clim = ERA5.groupby("valid_time.month").mean("valid_time")

# Anomalies
anom = ERA5.groupby("valid_time.month") - clim

# Calculate monthly linear trend at each grid point
trends = []
for month in [3, 4, 5]:
    anom_m = anom.where(anom["valid_time"].dt.month == month, drop=True)

    trend = (
        anom_m
        .polyfit(dim="year", deg=1)
        ["polyfit_coefficients"]
        .sel(degree=1)
    ) * 10   # K/decade

    trends.append(trend)

# Combine
trend_da = xr.concat(trends, dim="month")
trend_da = trend_da.assign_coords(month=[3, 4, 5])
```

```python
# 1x3 subplots with shared colorbar
fig, axes = plt.subplots(
    1, 3,
    figsize=(12, 10),
    subplot_kw={'projection': ccrs.NorthPolarStereo()},
    dpi=300,
    constrained_layout=True
)

months = ["(a) March", "(b) April", "(c) May"]

# Loop through each subplot
for i, ax in enumerate(axes):
    ax.set_extent([-180, 180, 54, 90], crs=ccrs.PlateCarree())

    # Filled SAT contours
    lats, lons = trend_da.latitude.values, trend_da.longitude.values
    trend_cyclic, lon_cyclic = add_cyclic_point(trend_da[i], coord=lons)
    im = ax.contourf(
        lon_cyclic, lats, trend_cyclic,
        cmap="RdBu_r",
        levels=np.linspace(-2, 2, 20),
        extend='both',
        transform=ccrs.PlateCarree()
    )

    ax.coastlines()
    ax.set_title(months[i], fontsize=18, loc='left')
    ax.set_boundary(circle_path, transform=ax.transAxes)

# Shared horizontal colorbar at bottom
cbar = fig.colorbar(
    im,
    ax=axes.ravel(),
    orientation='vertical',
    ticks=np.linspace(-2, 2, 9),
    shrink=0.4,
    aspect=15,
    pad=0.02
)

cbar.ax.tick_params(labelsize=12)
cbar.set_label("ERA5 SAT Anomaly Trend (K decade$^{-1}$)", fontsize=14)
plt.show()
```

### Figure 3 - Percent explained by winds

```python
def spatial_average(array, lats):
    weights = np.cos(np.deg2rad(lats))
    if len(array.shape) == 2:
        weights_array = weights[:, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0]*shape[1]))
        weighted_sum = np.nansum(weighted_reshaped, axis=0)
    if len(array.shape) == 3:
        weights_array = weights[np.newaxis, :, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0], shape[1]*shape[2]))
        weighted_sum = np.nansum(weighted_reshaped, axis=1)
    if len(array.shape) == 4:
        weights_array = weights[np.newaxis, np.newaxis, :, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0], shape[1], shape[2]*shape[3]))
        weighted_sum = np.nansum(weighted_reshaped, axis=2)
    if len(array.shape) == 5:
        weights_array = weights[np.newaxis, np.newaxis, np.newaxis, :, np.newaxis]
        weighted_array = np.multiply(array, weights_array)
        shape = np.shape(weighted_array)
        weighted_reshaped = np.reshape(weighted_array, (shape[0], shape[1], shape[2], shape[3]*shape[4]))
        weighted_sum = np.nansum(weighted_reshaped, axis=3)
    denom = np.nansum(weights) * shape[-1]
    return np.divide(weighted_sum, denom)


months = ["March", "April", "May"]
lats = nudged_trends.lat.values

nudged_monthly_arctic_trends = []
era5_monthly_arctic_trends   = []
obs_monthly_arctic_trends    = []

for i in range(3):
    print('---------- Nudged', months[i], '----------')

    Arctic = spatial_average(nudged_sat[i][170:], lats[170:])
    nudged_monthly_arctic_trends.append(Arctic)
    Arctic_spread = np.std(spatial_average(nudged_sat_ens[:, i, 170:], lats[170:]))
    print(f"Arctic:      {Arctic:.3f} +/- {2*Arctic_spread:.3f} K/dec")
    
    Global = spatial_average(nudged_sat[i], lats)
    Global_spread = np.std(spatial_average(nudged_sat_ens[:, i], lats))
    print(f"Global:      {Global:.3f} +/- {2*Global_spread:.3f} K/dec")

    print('\n---------- Others', months[i], '----------')

    ERA5 = spatial_average(trend_da[i][::-1].values[640:], trend_da[::-1].latitude.values[640:])
    era5_monthly_arctic_trends.append(ERA5)
    print(f"ERA5:        {ERA5:.3f} K/dec")

    Observed = spatial_average(obs_sat[i, 64:], lat[64:])
    obs_monthly_arctic_trends.append(Observed)
    print(f"My observed: {Observed:.3f} K/dec\n")

print('---------- Nudged Spring Means ----------')
spring_mean = np.mean(nudged_sat, axis=0)
print(f"Arctic Nudged Spring Mean: {spatial_average(spring_mean[170:], lats[170:]):.3f} K/dec")

observed_mean = spatial_average(era5_da[::-1].values[640:], era5_da[::-1].latitude.values[640:])
print(f"Arctic ERA5 Spring Mean:   {observed_mean:.3f} K/dec")
```

```python
# Observed values (x = SLP, y = SAT)
observed = [
    (-236.8718114595667, 0.32643064679727984),
    (-79.40412441449607, 0.8526703529858402),
    (-29.191823970081302, 0.4206435706147982),
]

months = ["March", "April", "May"]

def report_stats(x_avg, y_avg, obs_x, obs_y):
    # Means and standard deviations
    x_mean, x_std = np.mean(x_avg), np.std(x_avg, ddof=1)
    y_mean, y_std = np.mean(y_avg), np.std(y_avg, ddof=1)

    # Sigma distances
    x_sigma = (obs_x - x_mean) / x_std
    y_sigma = (obs_y - y_mean) / y_std
    hyp_sigma = np.sqrt(x_sigma**2 + y_sigma**2)

    # Percentages
    slp_pct = np.mean(x_avg <= obs_x) * 100
    sat_pct = np.mean(y_avg >= obs_y) * 100
    tot_pct = np.mean((x_avg <= (2/3)*obs_x) & (y_avg >= (2/3)*obs_y)) * 100

    return x_sigma, y_sigma, hyp_sigma, slp_pct, sat_pct, tot_pct


for i, (obs_x, obs_y) in enumerate(observed):
    xσ, yσ, hσ, slp_pct, sat_pct, tot_pct = report_stats(x_avgs[i], y_avgs[i], obs_x, obs_y)

    print(f"\n--------- {months[i]} ---------")
    print(f"Observed SLP:       {xσ:.1f}σ")
    print(f"Observed SAT:       {yσ:.1f}σ")
    print(f"Distance from mean: {hσ:.1f}σ")
    print(f"SLP percentile:     {slp_pct:.0f}%")
    print(f"SAT percentile:     {sat_pct:.0f}%")
    print(f"Joint percentile:   {tot_pct:.0f}%")

print("\n")
```

#### Pressure-latitude temperature trends

```python
# Load observed vertical temperature data
ds = xr.open_dataset('./data/ERA5_Vertical_Temp.nc').t

# Remove monthly climatology
def monthly_anomalies(da):
    clim = da.groupby('valid_time.month').mean('valid_time')
    return da.groupby('valid_time.month') - clim

# Apply anomalies to the dataset
ds_sel = ds.sel(valid_time=slice('1980-01-01', '2022-12-31'))
ds_anom = monthly_anomalies(ds_sel)

march = ds_anom.sel(valid_time=ds_anom.valid_time.dt.month == 3)
april = ds_anom.sel(valid_time=ds_anom.valid_time.dt.month == 4)
may   = ds_anom.sel(valid_time=ds_anom.valid_time.dt.month == 5)

# Select latitudes 60°N and poleward
lat_mask = ds_anom.latitude >= 60
march = march.sel(latitude=lat_mask)
april = april.sel(latitude=lat_mask)
may   = may.sel(latitude=lat_mask)

# Zonal average
march_zonal = march.mean(dim="longitude")
april_zonal = april.mean(dim="longitude")
may_zonal   = may.mean(dim="longitude")

# Convert valid_time to year
march_zonal_year = march_zonal.groupby('valid_time.year').mean('valid_time')
april_zonal_year = april_zonal.groupby('valid_time.year').mean('valid_time')
may_zonal_year   = may_zonal.groupby('valid_time.year').mean('valid_time')

# Calculate trend
n_time = march_zonal.sizes['valid_time']
time_years = np.linspace(1980, 2022, n_time)

def linear_trend_3d(data, time_years):
    ntime, nlat, nplev = data.shape
    slopes = np.full((nlat, nplev), np.nan)

    for i in range(nlat):
        for j in range(nplev):
            y = data[:, i, j]
            mask = ~np.isnan(y)
            if np.sum(mask) >= 2:
                slope, _ = np.polyfit(time_years[mask], y[mask], 1)
                slopes[i, j] = slope * 10
            else:
                slopes[i, j] = np.nan

    return slopes

obs_march_trend = linear_trend_3d(march_zonal, time_years)
obs_april_trend = linear_trend_3d(april_zonal, time_years)
obs_may_trend   = linear_trend_3d(may_zonal, time_years)

print('Done!')
```

```python
# Get SLP data
slp_paths = glob.glob('./data/obs-grids/SLP-regrid/*')

slp_list = []
for path in slp_paths:
    ds = xr.open_dataset(path).DATA / 100  # hPa
    slp_list.append(ds)

slp_all = xr.concat(slp_list, dim='dataset')
slp = slp_all.mean(dim='dataset')

# lat, lon for cut off area
latc, lonc = slp.lat.values, slp.lon.values

# Define region mask
lat_mask = (latc >= 35) & (latc <= 90)
lon_mask = (lonc >= 60) & (lonc <= 80)

# Select region and compute SLP index
slp_index = spatial_average(np.array(slp[:, :, lat_mask, lon_mask]), latc[lat_mask], len(lonc[lon_mask]))

labels = ["March", "April", "May"]
colors = ["Purple", "Grey", "Orange"]
plt.figure(figsize=(10, 3))
for i in range(3):
    plt.plot(slp_index[:, i], label=labels[i], color=colors[i])
plt.legend(frameon=False)
plt.grid(alpha=0.3)
plt.xlim(0, 42)
plt.ylim(-7, 7)
plt.xlabel('Years since 1980')
plt.ylabel('hPa')
plt.show()

slp_index.shape
```

```python
# Get SIC data
ds = xr.open_dataset('./data/G02202_25km_1978_2025.nc').cdr_seaice_conc

# Monthly means
ds_monthly = (ds.resample(time="MS").mean(skipna=True))

# Mask invalid values
ds_monthly = ds_monthly.where((ds_monthly >= 0) & (ds_monthly <= 1))

# March-May, 1980-2022
ds_monthly = (
    ds_monthly
    .sel(time=slice("1980-01-01", "2022-12-31"))
    .where(ds_monthly["time.month"].isin([3, 4, 5]), drop=True)
)

# Build 2D x/y grid (meters)
x2d, y2d = np.meshgrid(ds_monthly.x.values, ds_monthly.y.values)

# Coordinate transformer: NSIDC to lat/lon
transformer = Transformer.from_crs(
    "EPSG:3411",
    "EPSG:4326",
    always_xy=True
)

lon2d, lat2d = transformer.transform(x2d, y2d)

latg = xr.DataArray(
    lat2d,
    dims=("y", "x"),
    coords={"y": ds.y, "x": ds.x},
    name="lat"
)

ds_monthly = ds_monthly.assign_coords(lat=latg)

# Reshape into (years, months)
n_years, n_months = 43, 3

sic = ds_monthly.values.reshape(
    n_years, n_months,
    ds_monthly.sizes["y"],
    ds_monthly.sizes["x"],
)

sic.shape
```

```python
import warnings
warnings.filterwarnings("ignore", message="Input array is not C_CONTIGUOUS")

# Regrid SLP trend to SIC grid
ds_target = xr.Dataset(
    {
        "lat": (("y", "x"), lat2d),
        "lon": (("y", "x"), lon2d)
    }
)

# Create regridder
regridder = xe.Regridder(
    slp,        # source: cut lat/lon grid
    ds_target,  # target: NSIDC 25 km grid
    method="bilinear",
    periodic=True,
    reuse_weights=False
)

# Regrid SLP onto NSIDC grid
# Select 44: for 20N (cut area)
obs_slp_on_nsidc = regridder(obs_slp[:, 44:])

obs_slp_on_nsidc.shape
```

```python
# Regress SIC onto SLP
_, n_months, ny, nx = sic.shape
beta_sic_slp = np.zeros((n_months, ny, nx))

for m in range(n_months):
    # Predictor: SLP index for month m
    x = slp_index[:, m]
    x_anom = x - x.mean()

    # Predictand: SIC for month m
    y = sic[:, m]
    y_anom = y - y.mean()

    # SIC onto SLP index
    beta_sic_slp[m] = np.mean(x_anom[:, None, None] * y, axis=0) / np.mean(x_anom**2)

beta_sic_slp.shape
```

```python
# 1x3 subplots with shared colorbar
nsidc_crs = ccrs.Stereographic(
    central_latitude=90,
    central_longitude=-45,
    true_scale_latitude=70
)

fig, axes = plt.subplots(
    1, 3,
    figsize=(12, 10),
    subplot_kw={'projection': nsidc_crs},
    dpi=300,
    constrained_layout=True
)

labels = ["(a) March", "(b) April", "(c) May"]

X, Y = np.meshgrid(ds_monthly.x.values, ds_monthly.y.values)

# Loop through each subplot
for i, ax in enumerate(axes):
    ax.set_extent([-180, 180, 54, 90], crs=ccrs.PlateCarree())
    ax.set_facecolor('white')
    ax.add_feature(cfeature.LAND, facecolor='lightgrey')

    im = ax.contourf(
        X, Y,
        # beta_sic_slp[i]*obs_slp_on_nsidc[i]*100,
        -beta_sic_slp[i]*100,
        cmap="RdBu_r",
        extend='both',
        levels=np.linspace(-20, 20, 20),
    )

    ax.coastlines(resolution='50m', linewidth=0.5)
    ax.set_title(labels[i], fontsize=18, loc='left')
    ax.set_boundary(circle_path, transform=ax.transAxes)
    ax.set_aspect('equal')

# Shared horizontal colorbar at bottom
cbar = fig.colorbar(
    im,
    ax=axes.ravel(),
    orientation='vertical',
    ticks=np.linspace(-20, 20, 9),
    shrink=0.4,
    aspect=15,
    pad=0.02
)

cbar.ax.tick_params(labelsize=12)

# cbar.set_label("SIC (%) decade$^{-1}$", fontsize=14)
# plt.savefig('./figures/Figure6.png', dpi=300, bbox_inches='tight')

cbar.set_label("Change in SIC (%) per -1 hPa", fontsize=14)
plt.savefig('./figures/FigureS5.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Find r between spring-mean observations + regressed product SIC

# Calcualte observed SIC trend
nyear, nmonth, ny, nx = sic.shape
time = np.arange(nyear)

sic_trends = []
for m in range(3):
    sic_m = sic[:, m]
    sic_flat = sic_m.reshape(nyear, -1)
    slopes = np.array([np.polyfit(time, sic_flat[:, i], deg=1)[0] for i in range(sic_flat.shape[1])])
    sic_trends.append(slopes.reshape(ny, nx))

np.shape(sic_trends)
```

```python
# Seasonal mean
spring_obs = np.nanmean(sic_trends, axis=0)
spring_mult = np.nanmean(beta_sic_slp * obs_slp_on_nsidc, axis=0)

# Flattened arrays
spring_obs_flat = spring_obs.flatten()
spring_mult_flat = spring_mult.flatten()

# Mask NaNs
mask = ~np.isnan(spring_obs_flat) & ~np.isnan(spring_mult_flat)
spring_obs_clean = spring_obs_flat[mask]
spring_mult_clean = spring_mult_flat[mask]

# Pattern correlation
r = np.corrcoef(spring_obs_clean, spring_mult_clean)[0, 1]

print(f"r = {r:.2f}")
```

```python
from scipy.stats import pearsonr

# For each month
nmonths, ny, nx = np.shape(sic_trends)

# Flatten
sic_flat = np.reshape(sic_trends, (nmonths, -1))
slp_mult_flat = np.reshape((beta_sic_slp * obs_slp_on_nsidc), (nmonths, -1))

r_monthly = np.zeros(months)
p_monthly = np.zeros(months)

for m in range(months):
    sic_m = sic_flat[m, :]
    slp_m = slp_mult_flat[m, :]
    
    mask = ~np.isnan(sic_m) & ~np.isnan(slp_m)
    
    r_monthly[m], p_monthly[m] = pearsonr(sic_m[mask], slp_m[mask])

for m in range(months):
    if p_monthly[m] < 0.05:
        print(f"Month {m+1}: r = {r_monthly[m]:.2f}, p = {p_monthly[m]:.3f}")
```

#### SAT regression

```python
# Get SAT data
sat_paths = glob.glob('./data/obs-grids/SAT-regrid/*')

sat_list = []
for path in sat_paths:
    ds = xr.open_dataset(path).DATA
    sat_list.append(ds)

sat_all = xr.concat(sat_list, dim='dataset')
sat = sat_all.mean(dim='dataset')

lats, lons = ds.lat.values, ds.lon.values

# Regress SAT onto SLP
_, n_months, nlat, nlon = sat.shape
beta_sat_slp = np.zeros((n_months, nlat, nlon))

for m in range(n_months):
    # Predictor: SLP index for month m
    x = slp_index[:, m]
    x_anom = x - x.mean()

    # Predictand: SAT for month m
    y = sat[:, m]
    y_anom = y - y.mean(axis=0)

    # SAT onto SLP index
    beta_sat_slp[m] = np.mean(x_anom[:, None, None] * y_anom, axis=0) / np.mean(x_anom**2)

beta_sat_slp.shape
```
