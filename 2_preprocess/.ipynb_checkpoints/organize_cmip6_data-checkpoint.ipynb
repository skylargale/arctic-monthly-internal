```python
import numpy as np
import glob
import netCDF4 as nc
import xarray as xr
import xesmf as xe
import warnings
warnings.filterwarnings("ignore")
```

#### Functions

```python
def organizer(sat_and_slp_files, processing):
    sat_all_ensembles = []
    slp_all_ensembles = []

    for file_set in sat_and_slp_files:
        model_name = str(file_set[0].split('/')[6])
        member = str(file_set[0].split('/')[7])
        name = model_name+'/'+member
        sat_hist = xr.open_dataset(file_set[0], engine='netcdf4')
        slp_hist = xr.open_dataset(file_set[1], engine='netcdf4')

        try:
            if processing == 'spliced':
                num_years = 200
                num_months = num_years*12

                # open splicable files
                sat_ssp = xr.open_dataset(file_set[2], engine='netcdf4')
                slp_ssp = xr.open_dataset(file_set[3], engine='netcdf4')

                # splice along time axis
                sat_data = xr.concat([sat_hist, sat_ssp], dim='time')
                slp_data = xr.concat([slp_hist, slp_ssp], dim='time')

                # sort by time
                sat_data = sat_data.sortby('time')
                slp_data = slp_data.sortby('time')

                # select 1851 to 2050
                sat_data = sat_data.sel(time=slice('1851', '2050'))
                slp_data = slp_data.sel(time=slice('1851', '2050'))

                # close files
                sat_hist.close()
                slp_hist.close()
                sat_ssp.close()
                slp_ssp.close()

            if processing == 'hist':
                num_years = 164
                num_months = num_years*12

                # sort by time
                sat_data = sat_hist.sortby('time')
                slp_data = slp_hist.sortby('time')

                # select 1851 to 2014
                sat_data = sat_data.sel(time=slice('1851', '2014'))
                slp_data = slp_data.sel(time=slice('1851', '2014'))

                # close files
                sat_hist.close()
                slp_hist.close()
        except:
            print('Failed to open:', name)
            sat_hist.close()
            slp_hist.close()
            continue

        print('Successfully opened:', name)

        try:
            lats_in = sat_data.lat.values
            lons_in = sat_data.lon.values
        except:
            print('Failed, weird grid:', name)
            continue

        # make sure there aren't any duplicate times
        sat_data = sat_data.drop_duplicates(dim='time', keep='first')
        slp_data = slp_data.drop_duplicates(dim='time', keep='first')

        if len(sat_data.time.values) != num_months:
            print("Does not have appropriate number of months:", name)
            continue

        # regrid models
        InputGrid = {"lon": lons_in, "lat": lats_in}
        OutputGrid = {"lon": np.arange(1.25, 358.751, 2.5), "lat": np.arange(-88.75, 88.751, 2.5)}
        regridder = xe.Regridder(InputGrid, OutputGrid, "bilinear", periodic=True)
        sat_regridded = regridder(sat_data)['tas']
        slp_regridded = regridder(slp_data)['psl']

        # reshape into years, months
        shape = np.shape(sat_regridded)
        sat_new = np.reshape(sat_regridded.to_numpy(), (int(shape[0]/12), 12, shape[1], shape[2]))
        slp_new = np.reshape(slp_regridded.to_numpy(), (int(shape[0]/12), 12, shape[1], shape[2]))

        if len(sat_new) != num_years:
            print("Does not have appropriate number of years:", name)
            continue

        sat_all_ensembles.append(sat_new)
        slp_all_ensembles.append(slp_new)

    return (sat_all_ensembles, slp_all_ensembles, model_name)


def saver(combined_data, processing, model_name):
    shape = np.shape(combined_data)
    lons = np.arange(1.25, 358.751, 2.5)
    lats = np.arange(-88.75, 88.751, 2.5)
    years = np.arange(1851, 1851 + shape[1])
    months = np.arange(0, 12)

    # create a netcdf file
    if processing == 'spliced':
        filename = '/glade/work/skygale/new_mydata/spliced/'
    if processing == 'hist':
        filename = '/glade/work/skygale/new_mydata/hist/'

    fileName = filename + model_name + '.nc'
    dataset = nc.Dataset(fileName, 'w', format='NETCDF4')

    # define the dimensions
    dataset.createDimension('member', shape[0])  # number of members
    dataset.createDimension('year', shape[1])    # 164 or 200 (hist or ssp)
    dataset.createDimension('month', shape[2])   # 12
    dataset.createDimension('lat', shape[3])     # 72
    dataset.createDimension('lon', shape[4])     # 144
    dataset.createDimension('data', shape[5])    # 2

    # define the variables
    var = dataset.createVariable('SAT_SLP', np.float32, ('member', 'year', 'month', 'lat', 'lon', 'data'))
    var[:] = combined_data
    dataset.close()

    # open the NetCDF file in append mode
    dataset = nc.Dataset(fileName, 'a')

    members_var = dataset.createVariable('member', 'f8', ('member',))
    members_var[:] = np.arange(1, 1 + shape[0])

    year_var = dataset.createVariable('year', 'f8', ('year',))
    year_var[:] = years

    months_var = dataset.createVariable('month', 'f8', ('month',))
    months_var[:] = months

    lat_var = dataset.createVariable('lat', 'f8', ('lat',))
    lat_var[:] = lats

    lon_var = dataset.createVariable('lon', 'f8', ('lon',))
    lon_var[:] = lons

    tas_and_psl_var = dataset.createVariable('data', 'f8', ('data',))
    tas_and_psl_var[:] = [1, 2]

    dataset.close()
```

#### Collect, sort, and clean up model paths

```python
# paths to models
spliced_paths = glob.glob('/glade/work/skygale/raw_CMIP6/ssp_psl_month/*')
hist_paths = glob.glob('/glade/work/skygale/raw_CMIP6/hist_psl_month/*')

# split to get spliced model names
spliced_models = [spliced_paths[i].split('/')[6] for i in range(len(spliced_paths))]

# find historical only models
hist_only_paths = []
for path in hist_paths:
    model_name = path.split('/')[6]
    if model_name not in spliced_models:
        hist_only_paths.append(path)

# separate LENS and non LENS
spliced_LENS_names = ['ACCESS-ESM1-5', 'CanESM5', 'E3SM-2-0', 'IPSL-CM6A-LR',
                      'MIROC6', 'MPI-ESM1-2-LR']
hist_LENS_names = ['GISS-E2-1-G', 'GISS-E2-1-H', 'MPI-ESM1-2-HR', 'NorCPM1']

spliced_LENS_to_process = []
spliced_non_LENS_to_process = []
for path in spliced_paths:
    if path.split('/')[6] in spliced_LENS_names:
        spliced_LENS_to_process.append(path)
    else:
        spliced_non_LENS_to_process.append(path)
spliced_LENS_to_process = sorted(spliced_LENS_to_process)
spliced_non_LENS_to_process = sorted(spliced_non_LENS_to_process)

hist_LENS_to_process = []
hist_non_LENS_to_process = []
for path in hist_only_paths:
    if path.split('/')[6] in hist_LENS_names:
        hist_LENS_to_process.append(path)
    else:
        hist_non_LENS_to_process.append(path)
hist_LENS_to_process = sorted(hist_LENS_to_process)
hist_non_LENS_to_process = sorted(hist_non_LENS_to_process)

# remove EC-Earth3 in non_LENS_to_process
spliced_non_LENS_to_process.pop(12)
len(spliced_non_LENS_to_process), spliced_LENS_to_process
```

#### Organize data by processing type

```python
processing = 'hist'
LENS = False

if processing == 'spliced' and LENS is True:
    for path in spliced_LENS_to_process:
        print(path)
        slp_files = glob.glob(path+'/*')
        sat_and_slp_files = []
        for slp_ssp in sorted(slp_files):
            slp_hist = slp_ssp.replace('ssp', 'hist')
            sat_ssp = slp_ssp.replace('psl', 'sat')
            sat_hist = sat_ssp.replace('ssp', 'hist')
            sat_and_slp_files.append([sat_hist, slp_hist, sat_ssp, slp_ssp])

        # organize sat and slp maps
        sat_all_ensembles, slp_all_ensembles, model_name = organizer(sat_and_slp_files, processing)

        # combine and save data
        sat_data = np.array(sat_all_ensembles)
        slp_data = np.array(slp_all_ensembles)
        combined_data = np.moveaxis([sat_data, slp_data], 0, -1)
        print(np.shape(combined_data))  # EM, 200, 12, 72, 144, 2
        saver(combined_data, processing, model_name)
        print('Done with', model_name)

OthersAllEM = []
if processing == 'spliced' and LENS is False:
    for path in spliced_non_LENS_to_process:
        print(path)
        slp_files = glob.glob(path+'/*')
        sat_and_slp_files = []
        for slp_ssp in sorted(slp_files):
            slp_hist = slp_ssp.replace('ssp', 'hist')
            sat_ssp = slp_ssp.replace('psl', 'sat')
            sat_hist = sat_ssp.replace('ssp', 'hist')
            sat_and_slp_files.append([sat_hist, slp_hist, sat_ssp, slp_ssp])

        # organize sat and slp maps
        sat_all_ensembles, slp_all_ensembles, model_name = organizer(sat_and_slp_files, processing)

        # combine and save data
        sat_data = np.array(sat_all_ensembles)
        slp_data = np.array(slp_all_ensembles)
        combined_data = np.moveaxis([sat_data, slp_data], 0, -1)
        print(np.shape(combined_data))  # EM, 200, 12, 72, 144, 2
        OthersAllEM.append(combined_data)
    saver(np.concatenate(OthersAllEM, 0), processing, 'OthersAllEM')

if processing == 'hist' and LENS is True:
    for path in hist_LENS_to_process:
        print(path)
        slp_files = glob.glob(path+'/*')
        sat_and_slp_files = []
        for slp_hist in sorted(slp_files):
            sat_hist = slp_hist.replace('psl', 'sat')
            sat_and_slp_files.append([sat_hist, slp_hist])

        # organize sat and slp maps
        sat_all_ensembles, slp_all_ensembles, model_name = organizer(sat_and_slp_files, processing)

        # combine and save data
        sat_data = np.array(sat_all_ensembles)
        slp_data = np.array(slp_all_ensembles)
        combined_data = np.moveaxis([sat_data, slp_data], 0, -1)
        print(np.shape(combined_data))  # EM, 164, 12, 72, 144, 2
        saver(combined_data, processing, model_name)
        print('Done with', model_name)

OthersAllEM = []
if processing == 'hist' and LENS is False:
    for path in hist_non_LENS_to_process:
        print(path)
        slp_files = glob.glob(path+'/*')
        sat_and_slp_files = []
        for slp_hist in sorted(slp_files):
            sat_hist = slp_hist.replace('psl', 'sat')
            sat_and_slp_files.append([sat_hist, slp_hist])

        # organize sat and slp maps
        sat_all_ensembles, slp_all_ensembles, model_name = organizer(sat_and_slp_files, processing)

        # combine and save data
        sat_data = np.array(sat_all_ensembles)
        slp_data = np.array(slp_all_ensembles)
        combined_data = np.moveaxis([sat_data, slp_data], 0, -1)
        print(np.shape(combined_data))  # EM, 164, 12, 72, 144, 2
        OthersAllEM.append(combined_data)
    saver(np.concatenate(OthersAllEM, 0), processing, 'OthersAllEM')
print('Done!')
```
