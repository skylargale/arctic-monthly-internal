```python
import xarray as xr
import requests
import os

xr.set_options(display_style='html')
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

```python
def esgf_search(server="https://esgf-node.llnl.gov/esg-search/search",
                files_type="OPENDAP", local_node=True, project="CMIP6",
                verbose=False, format="application%2Fsolr%2Bjson",
                use_csrf=False, **search):
    client = requests.session()
    payload = search
    payload["project"] = project
    payload["type"] = "File"
    if local_node:
        payload["distrib"] = "false"

    payload["format"] = format
    offset = 0
    numFound = 10000
    all_files = []
    files_type = files_type.upper()
    while offset < numFound:
        payload["offset"] = offset
        url_keys = []
        for k in payload:
            url_keys += ["{}={}".format(k, payload[k])]

        url = "{}/?{}".format(server, "&".join(url_keys))
        r = client.get(url)
        r.raise_for_status()
        resp = r.json()["response"]
        numFound = int(resp["numFound"])
        resp = resp["docs"]
        offset += len(resp)
        for d in resp:
            url = d["url"]
            for f in d["url"]:
                sp = f.split("|")
                if sp[-1] == files_type:
                    file = sp[0].split(".html")[0]
                    new_file = file[:36]+'fileServer'+file[41:]
                    all_files.append(new_file)
    return sorted(all_files)
```

```python
# historical sat
for member in range(1, 22):
    member_string = str(member)
    member_identification = 'r' + member_string + 'i1p1f1'
    print(member_identification+'...')
    urls = esgf_search(activity_id='CMIP',
                       data_node='esgf-data2.llnl.gov',
                       table_id='Amon',
                       variable_id='tas',
                       experiment_id='historical',
                       institution_id="E3SM-Project",
                       source_id="E3SM-2-0",
                       member_id=member_identification
                       )

    # download locally
    download_dir = 'downloads'
    os.makedirs(download_dir, exist_ok=True)

    local_files = []
    for url in urls:
        local_filename = os.path.join(download_dir, os.path.basename(url))
        response = requests.get(url)
        with open(local_filename, 'wb') as f:
            f.write(response.content)
        local_files.append(local_filename)

    try:
        dataset = xr.open_mfdataset(local_files, combine='by_coords')
        ds = dataset.drop_vars(['time_bnds', 'lat_bnds', 'lon_bnds'])
    except Exception as e:
        print(f"Error opening local datasets: {e}")

    # save as combined .nc file
    ds.load().to_netcdf('/glade/work/skygale/rawdata/hist_sat_month/E3SM-2-0/{member_identification}'.format(member_identification=member_identification))
print('Done!')
```
